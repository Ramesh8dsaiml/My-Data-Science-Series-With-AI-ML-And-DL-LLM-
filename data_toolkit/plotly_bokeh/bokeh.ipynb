```python
import pandas as pd
import numpy as np

from bokeh.plotting import figure, show, output_notebook, output_file
from bokeh.models import ColumnDataSource, HoverTool
    
output_notebook()
```

```python
import bokeh.sampledata
print(dir (bokeh.sampledata))
```

```python
# make a own data sets :
df = pd.DataFrame({
    "Date": pd.to_datetime([
        "2026-01-01", "2026-01-02", "2026-01-03", "2026-01-04",
        "2026-01-05", "2026-01-06", "2026-01-07", "2026-01-08",
        "2026-01-09", "2026-01-10"
    ]),
    "Product": [
        "Laptop", "Phone", "Tablet", "Laptop", "Phone",
        "Watch", "Tablet", "Laptop", "Phone", "Watch"
    ],
    "Region": [
        "North", "South", "East", "West", "North",
        "South", "East", "West", "North", "South"
    ],
    "Sales": [80000, 50000, 30000, 90000, 60000, 20000, 35000, 85000, 55000, 25000],
    "Profit": [12000, 8000, 5000, 15000, 10000, 4000, 6000, 14000, 9000, 4500],
    "Customers": [20, 35, 15, 25, 40, 18, 22, 30, 38, 20]
})

df.head()
```

```python
# in plotly : fig= px.line(.............)
# in bokeh : p= figure (.....)
p = figure(
    title="Sales Trend",
    width=900,
    height=400
)

show(p)
```

```python
p = figure(
    title="Sales Trend",
    x_axis_type="datetime",
    width=800,
    height=400
)

p.line(
    df["Date"],
    df["Sales"],
    line_width=3
)

show(p)
```

```python
p = figure(
    title="Customers vs Sales",
    width=800,
    height=400
)

p.scatter(
    df["Customers"],
    df["Sales"],
    size=10
)

show(p)
```

```python
# circle :

p = figure(
    title="Customers vs Sales",
    width=800,
    height=400
)

p.circle(
    df["Customers"],
    df["Sales"],
    size=20,
    alpha=1
    
)

show(p)
```

```python
# vertical bar chart  :
product_sales = df.groupby("Product", as_index=False)["Sales"].sum()

products = product_sales["Product"].tolist()
sales = product_sales["Sales"].tolist()

p = figure(
    x_range=products,
    title="Product-wise Sales",
    width=800,
    height=400
)

p.vbar(
    x=products,
    top=sales,
    width=0.7
)

show(p)
```

```python
# h  bar 
p = figure(
    y_range=products,
    title="Product-wise Sales",
    width=800,
    height=400
)

p.hbar(
    y=products,
    right=sales,
    height=0.5
)

show(p)
```

```python
source = ColumnDataSource(df)

p = figure(
    title="Sales Analysis",
    width=800,
    height=400
)

p.scatter(
    x="Customers",
    y="Sales",
    source=source,
    size=10
)

hover = HoverTool(
    tooltips=[
        ("Product", "@Product"),
        ("Region", "@Region"),
        ("Customers", "@Customers"),
        ("Sales", "@Sales"),
        ("Profit", "@Profit"),
        ("Date","@Date")
    ]
)

p.add_tools(hover)

show(p)
```

```python
p = figure(
    title="Sales Trend",
    x_axis_type="datetime",
    width=800,
    height=400
)

p.line(
    df["Date"],
    df["Sales"],
    line_width=3,
    line_color="navy"
)

show(p)
```

```python
p = figure(
    title="Customers vs Sales",
    width=800,
    height=400
)

p.scatter(
    df["Customers"],
    df["Sales"],
    size=15,
    fill_color="green",     #customize the marke'
    line_color="black",
    alpha=0.7
)

show(p)
```

```python
# background + grid :

p = figure(
    title="Sales Trend",
    x_axis_type="datetime",
    width=800,
    height=400
)

p.line(
    df["Date"],
    df["Sales"],
    line_width=3
)

p.background_fill_color = "lightgray"
p.grid.grid_line_alpha = 0.3

show(p)
```

```python
# axis title :
p = figure(
    title="Customers vs Sales",
    width=800,
    height=400
)

p.scatter(
    df["Customers"],
    df["Sales"],
    size=10
)

p.xaxis.axis_label = "Customers"
p.yaxis.axis_label = "Sales"

# title customize 

p.title.text = "Customer vs Sales Analysis"
p.title.text_font_size = "20pt"


show(p)
```

```python
# lagend :
p = figure(
    title="Sales vs Profit",
    x_axis_type="datetime",
    width=800,
    height=400
)

p.line(
    df["Date"],
    df["Sales"],
    line_width=3,
    legend_label="Sales"
)

p.line(
    df["Date"],
    df["Profit"],
    line_width=3,
    legend_label="Profit"

)
(p.scatter(
    df["Customers"],
    df["Region"],
    line_width=6,
    legend_label="sales",
    color="black"
))

p.legend.location = "top_left"

show(p)
```
