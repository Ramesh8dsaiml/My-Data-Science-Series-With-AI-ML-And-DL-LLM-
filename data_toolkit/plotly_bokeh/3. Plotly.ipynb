```python
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go

df = pd.DataFrame({
    "Date": pd.to_datetime([
        "2026-01-01", "2026-01-02", "2026-01-03", "2026-01-04",
        "2026-01-05", "2026-01-06", "2026-01-07", "2026-01-08",
        "2026-01-09", "2026-01-10"
    ]),
    "Product": [
        "Laptop", "Phone", "Tablet", "Laptop", "Phone",
        "Watch", "Tablet", "Laptop", "Phone", "Watch"
    ],
    "Category": [
        "Electronics", "Electronics", "Electronics", "Electronics",
        "Electronics", "Accessories", "Electronics", "Electronics",
        "Electronics", "Accessories"
    ],
    "Region": [
        "North", "South", "East", "West", "North",
        "South", "East", "West", "North", "South"
    ],
    "Sales": [80000, 50000, 30000, 90000, 60000, 20000, 35000, 85000, 55000, 25000],
    "Profit": [12000, 8000, 5000, 15000, 10000, 4000, 6000, 14000, 9000, 4500],
    "Customers": [20, 35, 15, 25, 40, 18, 22, 30, 38, 20]
})

df
```

```python
# using px:
fig = px.line( # line chart useing for trand / Time 
    df,
    x="Date",
    y="Sales",
    title="sales trands"
)
fig.show()
```

```python
df.columns
```

```python
# bar charts : category comparision 
fig = px.bar(
    df,
    x="Product",
    y="Sales",
)
fig.show()
```

```python
# horizontal bar charts :
fig = px.bar(
    df,
    x="Sales",
    y="Product",
    orientation="h",
    title="Product-wise Sales"
)

fig.show()
```

```python
# scatter plots : show the relationship ,means that , numerical vs numerical 
fig = px.scatter(
    df,
    x="Customers",
    y="Sales",
    title="Customers vs Sales"
    
)

fig.show()
```

```python
# histogram : data distribution 
fig = px.histogram(
    df,
    x="Sales",
    title="Sales Distribution"
)

fig.show()
```

```python
# bos plot:      usinng for median + spread +outliers
fig = px.box(
    df,
    x="Category",
    y="Sales",
    title="Sales Distribution by Category"
)

fig.show()
```

```python
# violin plots: distribution shape 

fig = px.violin(
    df,
    x="Region",
    y="Sales",
    title="Sales Distribution by Region",
)

fig.show()
```

```python
# pie charts : pie = share , proportion 
region_sales = df.groupby("Region", as_index=False)["Sales"].sum()

fig = px.pie(
    region_sales,
    names="Region",
    values="Sales",
    title="Sales Share by Region"
)

fig.show()
```

```python
# area charts :  Area = Trend + magnitude 
fig = px.area(
    df,
    x="Date",
    y="Sales",
    title="Sales Area Chart"
)

fig.show()
```

```python
# funnel charts :  using for process and stage 
funnel_df = pd.DataFrame({
    "Stage": ["Visitors", "Product View", "Add to Cart", "Purchase"],
    "Count": [1000, 700, 400, 200]
})

fig = px.funnel(
    funnel_df,
    x="Count",
    y="Stage",
    title="Sales Funnel"
)

fig.show()
```

```python
# heatmap :  using for metrix and correlations 
corr = df[["Sales", "Profit", "Customers"]].corr()

fig = px.imshow(
    corr,
    text_auto=True,
    title="Correlation Heatmap"
)

fig.show()
```

```python
# 3d scatter : 3 numerical variable 
fig = px.scatter_3d(
    df,
    x="Customers",
    y="Sales",
    z="Profit",
    color="Region",
    title="3D Sales Analysis"
)

fig.show()
```

```python
# colors: means extra information 

fig = px.scatter(
    df,
    x="Customers",
    y="Sales",
    color="Region",
    size="Profit",
    title="Sales by Region"
)

fig.show()
```

```python
fig = px.scatter(
    df,
    x="Customers",
    y="Sales",
    color="Region",
    symbol="Category",
    title="Sales Analysis"
)

fig.show()
```

```python
fig.update_layout(
    width=1100,
    height=600
)

fig.show()
```

```python
fig.update_layout(
    title={
        "text": "Monthly Sales Report",
        "x": 0.5
    }
)

fig.show()
```

```python
# Axis Labels :

fig.update_layout(
    xaxis_title="Number of Customers",
    yaxis_title="Sales Amount"
)

fig.show()
```

```python
# Axis Range 
fig.update_yaxes(
    range=[0, 100000]
)

fig.show()
```

```python
# Rotaete   X- Lable 
fig.update_xaxes(
    tickangle=15
)

fig.show()
```

```python
# # common Tamplate :
# plotly
# plotly_white
# plotly_dark
# ggplot2
# seaborn
# simple_white

fig.update_layout(
    template="plotly_dark"
    
)

fig.show()
```

```python
    # line /marker customize

fig.update_traces(
    mode="lines+markers",
    marker=dict(size=10),
    line=dict(width=2)
)

fig.show()
```

```python
# Legend Customize :  if you are hide  than use False 

fig.update_layout(
    legend_title="Region",
    showlegend=False
)

fig.show()
```

```python
# Annotation :  Highlight the important Point :

fig.add_annotation(
    x=df["Date"].iloc[-1],
    y=df["Sales"].iloc[-1],
    text="Highest Recent Sales",
    showarrow=True
)

fig.show()
```
