```python
# use lagend:
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr"]

sales_2025 = [100, 150, 130, 180]
sales_2026 = [120, 170, 160, 210]

plt.plot(months, sales_2025, label="2025 Sales")
plt.plot(months, sales_2026, label="2026 Sales")

plt.xlabel("Month")
plt.ylabel("Sales")
plt.title("Sales Comparison")

plt.legend()

plt.show()
```

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr"]

sales_2025 = [100, 150, 130, 180]
sales_2026 = [120, 170, 160, 210]

plt.plot(months, sales_2025, label="2025 Sales")
plt.plot(months, sales_2026, label="2026 Sales")

plt.xlabel("Month")
plt.ylabel("Sales")
plt.title("Sales Comparison")

plt.legend()
plt.grid(axis="x")
plt.grid(axis="y")


plt.show()
```

```python
products = ['A','B','C']
sales = [120,180,150]
plt.bar(products, sales)
plt.xlabel('Product')
plt.ylabel('Sales')
plt.title('Product Sales')
plt.show()
```

```python
hours = [1,2,3,4,5]
scores = [40,45,55,70,80]
plt.scatter(hours, scores)
plt.xlabel('Study Hours')
plt.ylabel('Score')
plt.title('Study Hours vs Score')
plt.show()
```
