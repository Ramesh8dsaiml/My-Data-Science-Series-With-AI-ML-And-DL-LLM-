# 4. Explain the use of the groupby() method in Pandas
### The groupby() method in Pandas is used to split data into groups based on one or more columns and apply aggregation functions like sum, mean, count, etc., to analyze grouped data.

### groupby() in Pandas?

- The groupby() method in Pandas is used to group data based on one or more columns and perform aggregate operations on those groups.
- It follows the concept:
- Split → Apply → Combine

```python
# example :
import pandas as pd

data = {
    "Name": ["A", "B", "C", "D", "E"],
    "Department": ["IT", "HR", "IT", "HR", "IT"],
    "Salary": [50000, 40000, 60000, 45000, 70000]
}

df = pd.DataFrame(data)
print(df)
```

```python
df.groupby("Department")["Salary"].mean()
```

```python
df.groupby("Department")["Salary"].agg(["mean", "sum", "count"])
```

# 19 . How does Pandas simplify time series analysis
## What is Time Series Analysis?
- Time series data is data indexed in a sequence over time.
#### Example: stock prices, sales data, temperature readings
- Pandas provides easy-to-use features for time series.
### Pandas simplifies time series analysis by providing datetime indexing, resampling, rolling window functions, and easy plotting for sequential time-based data.

# Pandas Features for Time Series
- DatetimeIndex for date-based indexing
- resample() to aggregate data over time intervals
- shift() for lag/lead operations
- rolling() for moving averages and window functions
- to_period() for period conversion


```python
import pandas as pd

dates = pd.date_range('2026-01-01', periods=5, freq='D')
data = pd.Series([100,120,130,150,170], index=dates)

data.resample('2D').sum()
data.rolling(window=3).mean()
```

# 21. Why is NumPy’s array slicing faster than Python’s list slicing
## NumPy’s array slicing is faster than Python list slicing because arrays are stored in contiguous memory, allowing direct access, whereas lists store elements with pointers, requiring element-by-element copying.

### Why NumPy slicing is faster
- NumPy arrays are stored in contiguous memory blocks
- Python lists store elements in non-contiguous memory (pointers)
- Slicing NumPy arrays uses direct memory access → very fast
- List slicing requires copying each element → slower



```python
import numpy as np
import time

# NumPy array
arr = np.arange(1000000)
start = time.time()
slice_arr = arr[1000:2000]
end = time.time()
print("NumPy slice time:", end-start)

# Python list
lst = list(range(1000000))
start = time.time()
slice_lst = lst[1000:2000]
end = time.time()
print("List slice time:", end-start)
```
