# What is Statistics?
### Statistics is the science of:

- Collecting data
- Organizing data
- Analyzing data
- Drawing conclusions

# Types of Statistics
1. Descriptive Statistics
- Summarizes data.
2. Inferential Statistics
- Makes predictions about population using sample.
