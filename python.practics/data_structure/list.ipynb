```python
# enumerate() Function:  adds a counter to an iterable and returns both the index and the element.
```

```python
fruits = ["Apple", "Banana", "Mango"]

for index, value in enumerate(fruits):
    print(index, value)
```

```python
# Functional Programming Functions in List: 
# 1 map():  applies a function to every element of an iterable.

numbers = [1, 2, 3, 4]

result = list(map(lambda x: x * 2, numbers))

print(result)
```

```python
numbers = [1, 2, 3, 4, 5]

squares = list(map(lambda x: x ** 2, numbers))

print(squares)
```

```python
names = ["aman", "rahul", "rohit"]

result = list(map(str.upper, names))

print(result)
```

```python
# filter(): returns elements that satisfy a condition.

numbers = [1, 2, 3, 4, 5, 6]

result = list(filter(lambda x: x % 2 == 0, numbers))

print(result)
```

```python
marks = [35, 78, 90, 42, 65]

result = list(filter(lambda x: x > 50, marks))

print(result)
```

```python
# reduce():  repeatedly applies a function to reduce an iterable to a single value.

# from functools import reduce

# reduce(function, iterable)

from functools import reduce

numbers = [1, 2, 3, 4, 5]

result = reduce(lambda x, y: x + y, numbers)

print(result)
```

```python
from functools import reduce

numbers = [1, 2, 3, 4]

result = reduce(lambda x, y: x * y, numbers)

print(result)
```

```python
from functools import reduce

numbers = [10, 50, 20, 40]

result = reduce(lambda x, y: x if x > y else y, numbers)

print(result)
```

```python
# zip() combines corresponding elements from multiple iterables. 
names = ["Aman", "Rahul", "Rohit"]
marks = [80, 90, 75]

result = list(zip(names, marks))

print(result)
```

```python
a = [1, 2, 3]
b = [10, 20, 30]

result = [x + y for x, y in zip(a, b)]

print(result)
```

```python
a = [1, 2, 3]
b = [10, 20]

print(list(zip(a, b)))
```

# map() vs filter() vs reduce():
| Function      | Purpose                                | Output                      |
| ------------- | -------------------------------------- | --------------------------- |
| `map()`       | Har element ko transform karta hai     | Multiple transformed values |
| `filter()`    | Condition se elements select karta hai | Selected values             |
| `reduce()`    | Values ko combine karta hai            | Usually one value           |
| `zip()`       | Iterables ke elements pair karta hai   | Tuples                      |
| `enumerate()` | Index + value deta hai                 | Pairs                       |
