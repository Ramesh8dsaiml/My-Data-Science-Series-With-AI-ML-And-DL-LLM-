# Modern F-Strings: 
F-strings provide a concise syntax for alignment using the format specification mini-language (<, >, ^)

1. < : Left-aligns the text within the given width. 

2. >: Right-aligns the text within the given width.

3. ^ : Center-aligns the text within the given width

````python
text = "Python"

print(f"'{text:<10}'")  # Left-aligned:   'Python    '
print(f"'{text:>10}'")  # Right-aligned:  '      Python'
print(f"'{text:^10}'")  # Center-aligned: '  Python  '
````

````python
# Padding with Custom Characters
print(f"{text:-<10}")  # Output: Python----
print(f"{text:*>10}")  # Output: ****Python
print(f"{text:.^10}")  # Output: ..Python..
````

# Built-in String Methods:
ljust(width, fillchar): Left-aligns text and pads the right side.

rjust(width, fillchar): Right-aligns text and pads the left side.

center(width, fillchar): Centers text and pads both sides evenly

````python
#Alignment & Padding:it is used to align the string in a specific way and pad it with spaces or other characters.
#center(): it will center the string in the specified width and pad it with spaces on both sides.
"Python".center(20)
````

# String Concatenation


**English:** Combining two or more strings using the `+` operator is called **String Concatenation**.

```python
first = "Data"
second = "Science"

result = first + " " + second

print(result)
```

Output:

```text
Data Science
```

### Important Point

`+` strings ko combine karta hai, lekin string aur number ko directly concatenate nahi kar sakte.

```python
name = "Ramesh"
age = 22

print(name + age)   # TypeError
```

Number ko string me convert karna hoga:

```python
print(name + " " + str(age))
```

Output:

```text
Ramesh 22
```

**English:** The `+` operator can concatenate strings, but you cannot directly concatenate a string with an integer or another incompatible data type. Use `str()` to convert the value first.


````python
name = "ramesh "
nikname= "kumar"
print(name+nikname) # when both varaible are char than it combine the string :
````
