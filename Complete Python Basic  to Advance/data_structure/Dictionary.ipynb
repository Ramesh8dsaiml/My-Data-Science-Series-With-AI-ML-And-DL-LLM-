```python
# Dictionary Comprehension:
# Dictionary comprehension is a concise way to create a dictionary using an iterable.:
# syntax: {key: value for item in iterable}

even = {x: x**2 for x in range(1, 20) if x % 2 == 0}

print(even)
```

```python
odd = {x: x**2 for x in range(1, 11) if x % 2 != 0}

print(even)
```

```python
# with if-else
result = {
    x: "Even" if x % 2 == 0 else "Odd"
    for x in range(1, 20)
}

print(result)
```

```python
# Creating Dictionary from Two Lists:
# zip() can combine corresponding elements of two iterables, and dict() can convert them into key-value pairs.

keys = ["name", "age", "marks"]
values = ["Rahul", 21, 85]

student = dict(zip(keys, values))

print(student)
```

```python
# Dictionary Iteration with Index / Counter: note-> dictionary dont have directly indexing but use the enumerate() and get the dictionary indexs :

student = {
    "name": "Rahul",
    "age": 21,
    "marks": 85
}

for index, (key, value) in enumerate(student.items()): 
    print(index, key, value)
```

```python
# Swap Keys and Values :
data = {
    "a": 1,
    "b": 2,
    "c": 3
}

reverse = {value: key for key, value in data.items()}

print(reverse)
```

```python
# sorting the dictionary : accourding to keys:
data = {
    "banana": 3,
    "apple": 5,
    "orange": 2
}

result = dict(sorted(data.items()))

print(result)
```

```python
result = dict(sorted(data.items(), key=lambda item: item[1])) # accourding to values :

print(result)
```

```python
# dessinding orders :
result = dict(
    sorted(data.items(), key=lambda item: item[1], reverse=True)
)
```
