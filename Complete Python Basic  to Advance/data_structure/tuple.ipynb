```python
# tuple packing :Tuple packing means grouping multiple values into a single tuple.
t = 10, 20, 30
t
```

```python
# tuple unpacking : Tuple unpacking assigns tuple elements to individual variables.
t = (10, 20, 30)

a, b, c = t

print(a)
print(b)
print(c)
```

```python
# Extended unpacking allows one variable with * to collect multiple remaining values.

t = (10, 20, 30, 40, 50,24)

a, *b, c = t

print(a)   # 10
print(b)   
print(c)   
```

```python
# Looping Through Tuple : A for loop can be used to iterate through tuple elements.
t= (1,2,3,4,5,6,7)
for x in t:
    print(x)
```

```python
t = ("A", "B", "C")

for index, value in enumerate(t):
    print(index, value)
```

```python
t = ("A", "B", "C")

for index, value in enumerate(t, start=0): # bydefault 0 
    print(index, value)
```
