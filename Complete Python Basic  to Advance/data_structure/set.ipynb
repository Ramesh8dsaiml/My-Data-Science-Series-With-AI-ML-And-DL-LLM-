```python
# Set Comprehension: same aslist but list use [],and set use {} otherwise  all about same in ComprehensioComprehensio
# Set comprehension is a concise way to create a set from an iterable.
# basic syntax: {expression for item in iterable}

#  normal Loop:

numbers = {1, 2, 3, 4, 5}

squares = set()

for x in numbers:
    squares.add(x ** 2)

print(squares)
```

```python
# set comprehesion:


numbers = {1, 2, 3, 4, 5}

squares = {x ** 2 for x in numbers}

print(squares)
```

```python
# with if comdition:
# syntax: {expression for item in iterable if condition}

numbers = range(1, 11)

even = {x for x in numbers if x % 2 == 0}

print(even) 
```

```python
numbers = range(1, 41)

odd = {x for x in numbers if x % 2 !=0}

print(odd)
```

```python
names = ["Aman", "Rahul", "Aman", "Rohit"]

result = {name.upper() for name in names}

print(result)
```

```python
numbers = range(1, 40)

result = {
    "Even" if x % 2 == 0 else "Odd"
    for x in numbers
}

print(result)
```
